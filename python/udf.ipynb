```python
def calculate(a, b):
    addition = a + b
    subtraction = a - b

    return addition, subtraction

x, y = calculate(20, 10)
print(f"Sum :- {x}")
print(f"Sub :- {y}")
```

```python
def find_sum(no):
    total = 0;
    for number in  no:
        total += number
        
    return total

numbers = [10,20,30,40,50]
print(find_sum(numbers))
```

```python
def student_details(st):
    for key , value in st.items():
        print(f"{key} : {value}")

student={
    "name":"ajay",
    "age" : 21
}

student_details(student)
```
