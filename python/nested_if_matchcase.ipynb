```python
#nested if else
a=100;b=55;c=11
if a > b:
    if a > c:
        print("A is Max")
    else:
        print("C is Max")
else:
    if b > c:
        print("B is Max")
    else:
        print("C is Max")
```

```python
no=5
if no > 0:
    if no % 2 == 0:
        print("Even no")
    else:
        print("Odd no")
else:
    if no == 0:
        print("No negative or no positive")
    else:
        print("No is negative")
```

```python
#match case
day=8

match day:
    case 1:
        print("Monday")
    case 2:
        print("Tuesday")

    case _:
        print("Invalid Value")
```
